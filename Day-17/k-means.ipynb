```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

```

```python
x,y=make_blobs(n_samples=1000,centers=3,n_features=2,cluster_std=0.60,random_state=0)
```

```python
x
```

```python
y
```

```python
plt.scatter(x[:,0],x[:,1],c=y,cmap='viridis')
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
scaler=StandardScaler()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.33,random_state=42)    
```

```python
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++')
    kmeans.fit(x_train_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1, 11), wcss)
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++')
kmeans.fit_predict(x_train_scaled)
```

```python
y_pred = kmeans.predict(x_test_scaled)
y_pred
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_coefficient =[]
for k in range(2,11):
    kmeans = KMeans(n_clusters=k, init='k-means++')
    kmeans.fit(x_train_scaled)
    y_kmeans = kmeans.predict(x_train_scaled)
    score = silhouette_score(x_train_scaled, y_kmeans)
    silhouette_coefficient.append(score)
```

```python
silhouette_coefficient
```

```python
plt.plot(range(2,11),silhouette_coefficient)
plt.title('Silhouette Coefficient Method')
plt.xlabel('Number of clusters')        
plt.ylabel('Silhouette Coefficient')
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_moons, make_circles
import matplotlib.pyplot as plt
```

```python
x_moon, y_moon  = make_moons(n_samples=300, noise=0.1)
```

```python
x_circle , y_circle = make_circles(n_samples=300, noise=0.05, factor=0.1)
```

```python
plt.figure(figsize=(12, 5))
plt.scatter(x_moon[:, 0], x_moon[:, 1], c=y_moon)
plt.scatter(x_circle[:, 0], x_circle[:, 1], c=y_circle)
```

```python
from sklearn.preprocessing import StandardScaler
x_scaled_circle = StandardScaler().fit_transform(x_circle)
x_scaled_moon = StandardScaler().fit_transform(x_moon)
dbscan_circle = DBSCAN(eps=0.3, min_samples=5)
dbscan_moon = DBSCAN(eps=0.3, min_samples=5)
dbscan_circle.fit(x_scaled_circle)
dbscan_moon.fit(x_scaled_moon)
```

```python
dbscan_circle.labels_
```

```python
dbscan_moon.labels_
```

```python
plt.scatter(x_scaled_circle[:, 0], x_scaled_circle[:, 1], c=dbscan_circle.labels_)

```
